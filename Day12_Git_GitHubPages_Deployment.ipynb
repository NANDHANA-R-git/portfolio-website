## 5. Useful Git Commands Reference

| Command | Purpose |
|---|---|
| `git init` | Initialize a new Git repository |
| `git clone <url>` | Clone an existing repository |
| `git status` | Check current changes/state |
| `git add <file>` / `git add .` | Stage specific file / all changes |
| `git commit -m "message"` | Save staged changes with a message |
| `git push origin main` | Upload commits to GitHub |
| `git pull origin main` | Fetch and merge latest changes |
| `git log --oneline` | View commit history |
| `git branch` | List branches |
| `git checkout -b <branch>` | Create and switch to a new branch |
| `git merge <branch>` | Merge a branch into current branch |

## 6. Deploying with GitHub Pages

Steps followed to deploy the portfolio website:

1. Go to the GitHub repository → **Settings** → **Pages**.
2. Under **Build and deployment → Source**, select **Deploy from a branch**.
3. Choose the **`main`** branch and the **`/root`** folder (or `/docs` if using that convention).
4. Click **Save**.
5. GitHub generates a live URL in the format:

```
https://<username>.github.io/<repository-name>/
```

6. Wait 1–2 minutes for the deployment to complete, then verify the live site loads correctly.

## Form validation with JS

```html
<form id="contactForm" novalidate>
  <label for="name">Name</label>
  <input type="text" id="name" name="name">
  <span class="error-msg" id="nameError"></span>

  <label for="email">Email</label>
  <input type="email" id="email" name="email">
  <span class="error-msg" id="emailError"></span>

  <label for="message">Message</label>
  <textarea id="message" name="message" rows="5"></textarea>
  <span class="error-msg" id="messageError"></span>

  <button type="submit" class="btn">Send</button>
</form>
```

Added `novalidate` on the form so the browser's default popups don't fire - handling all the validation and messaging myself instead.

```javascript
const form = document.getElementById('contactForm');

form.addEventListener('submit', (e) => {
  e.preventDefault();
  let isValid = true;

  const name = document.getElementById('name');
  const email = document.getElementById('email');
  const message = document.getElementById('message');

  const nameError = document.getElementById('nameError');
  const emailError = document.getElementById('emailError');
  const messageError = document.getElementById('messageError');

  nameError.textContent = '';
  emailError.textContent = '';
  messageError.textContent = '';

  if (name.value.trim() === '') {
    nameError.textContent = 'Please enter your name.';
    isValid = false;
  }

  const emailPattern = /^[^\s@]+@[^\s@]+\.[^\s@]+$/;
  if (!emailPattern.test(email.value.trim())) {
    emailError.textContent = 'Please enter a valid email address.';
    isValid = false;
  }

  if (message.value.trim() === '') {
    messageError.textContent = 'Please enter a message.';
    isValid = false;
  }

  if (isValid) {
    alert('Message sent! (demo only, not actually connected yet)');
    form.reset();
  }
});
```

## Styling the error states

```css
.error-msg {
  color: #d92d20;
  font-size: 0.85rem;
  min-height: 18px;
  display: block;
}
```

Gave the error span a `min-height` even when empty, otherwise the form jumps around every time an error appears or disappears.

## Responsiveness testing

Opened Chrome DevTools (`Ctrl+Shift+I` / `Cmd+Option+I`) and used the device toolbar to test on presets like iPhone SE, iPhone 12 Pro, iPad Air, and a generic desktop size, rather than just resizing the window like I did on Day 6.

Caught a couple of small things this way:
- the projects grid felt a little tight on iPad-sized screens, added a bit more gap
- the scroll-to-top button overlapped the footer text on short mobile screens, adjusted its bottom offset

## Bug fix: CSS/JS file paths after deployment

After deploying with GitHub Pages on Day 12, the site loaded but with **no styling and no interactivity** - just plain unstyled HTML. Turned out the uploaded `style.css` and `script.js` ended up sitting at the root of the repo instead of inside `css/` and `js/` folders like `index.html` was expecting.

```html
<!-- before - pointing to folders that didn't actually exist on the deployed repo -->
<link rel="stylesheet" href="css/style.css">
<script src="js/script.js"></script>

<!-- after - pointing to where the files actually are -->
<link rel="stylesheet" href="style.css">
<script src="script.js"></script>
```

Good reminder that what you *think* got uploaded and what actually landed in the repo can be two different things - always worth checking the live file structure after deploying, not just locally.

Codomax Web Dev Internship

Today was all about getting the basic structure of the portfolio page going. Started with plain HTML, no styling yet - that's for Day 4. Just wanted to get the Home, Hero and About sections in place with proper tags.


## What I did today

- went through HTML basics again (tags, structure, attributes)
- set up the main index.html file
- added the Hero section - name, short tagline, a button
- added the About section - a small bio and a spot for my photo

Nothing fancy, just plain HTML for now.


## Quick notes on tags I used

`<header>` - top part of the page

`<section>` - used this to separate Hero and About instead of just divs everywhere, feels more organized

`<h1>` for the main heading, kept it to one per page

`<p>` for text

`<img>` for the profile pic (just a placeholder path for now)

`<a>` for the button/link

Gave each section an `id` too since I'll need those for the nav links later on.


## index.html so far

```html
<!DOCTYPE html>
<html lang="en">
<head>
  <meta charset="UTF-8">
  <meta name="viewport" content="width=device-width, initial-scale=1.0">
  <title>My Portfolio</title>
  <link rel="stylesheet" href="css/style.css">
</head>
<body>

  <section class="hero" id="home">
    <h1>Hi, I'm Nandhana</h1>
    <p>Web developer in progress, still figuring things out one day at a time.</p>
    <a href="#contact" class="btn">Get In Touch</a>
  </section>

  <section class="about" id="about">
    <h2>About Me</h2>
    <img src="assets/images/profile.jpg" alt="my photo">
    <p>
      I'm a 7th sem ECE student learning web development on the side.
      Right now I'm just getting comfortable with HTML and CSS before
      moving on to JS later this internship.
    </p>
  </section>

  <script src="js/script.js"></script>
</body>
</html>
```

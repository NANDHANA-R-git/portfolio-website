# Day 5 - Responsive Layouts with Flexbox & CSS Grid

Codomax Web Dev Internship

Today was about making the layout actually flexible instead of everything just stacking top to bottom. Used Flexbox for the nav bar and smaller layout bits, and CSS Grid for the Skills section so the items line up in a neat grid instead of a plain list.


## What I did today

- converted the nav bar to Flexbox properly (was already close, cleaned it up)
- used Flexbox on the Hero content to center things both ways
- switched the Skills section from a `<ul>` list to a CSS Grid layout
- started thinking about how this will need to collapse on smaller screens (media queries are tomorrow though)


## Flexbox vs Grid - how I'm thinking about it

Still getting a feel for when to use which, but roughly:

- Flexbox for one-directional stuff - nav links in a row, centering content in the hero
- Grid for anything that needs rows AND columns - like the skills cards, and later the projects section

Not a strict rule but it's helping me decide instead of guessing.


## Nav bar with Flexbox

```css
nav {
  display: flex;
  justify-content: center;
  align-items: center;
  gap: 24px;
  padding: 16px;
}
```

Nothing changed structurally, just cleaner alignment now.


## Hero section centered with Flexbox

```css
.hero {
  display: flex;
  flex-direction: column;
  align-items: center;
  justify-content: center;
  text-align: center;
  min-height: 60vh;
  padding: 40px 20px;
}
```

Using `min-height: 60vh` so the hero takes up a good chunk of the screen without hardcoding a pixel value.


## Skills section as a Grid

Changed the HTML from a `<ul>` to divs so I could treat each skill as a card:

```html
<section class="skills" id="skills">
  <h2>Skills</h2>
  <div class="skills-grid">
    <div class="skill-card">Python</div>
    <div class="skill-card">VLSI</div>
    <div class="skill-card">LTspice</div>
    <div class="skill-card">Leadership</div>
    <div class="skill-card">Public Speaking</div>
    <div class="skill-card">Project Management</div>
  </div>
</section>
```

```css
.skills-grid {
  display: grid;
  grid-template-columns: repeat(3, 1fr);
  gap: 16px;
  max-width: 800px;
  margin: 0 auto;
}

.skill-card {
  background-color: var(--section-bg);
  border: 1px solid #e5e5e5;
  border-radius: 8px;
  padding: 20px;
  text-align: center;
  font-weight: 500;
  transition: transform 0.2s ease;
}

.skill-card:hover {
  transform: translateY(-4px);
}
```

3 columns for now on desktop, will need to drop to fewer columns on smaller screens tomorrow with media queries. Planning to reuse this same grid pattern for the Projects section later on Day 8.

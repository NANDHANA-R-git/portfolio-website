## Notes on semantic HTML

Turns out using the right tag actually matters, not just for looks but for accessibility and SEO too. Switched some generic divs to proper tags:

`<nav>` for the navigation bar

`<main>` to wrap the core page content

`<section>` for skills, education, contact - same as before

`<footer>` for the bottom bit

`<form>`, `<label>`, `<input>`, `<textarea>` for the contact form

`<ul>` / `<li>` for the skills list instead of just separate p tags

Screen readers apparently rely a lot on these tags to figure out page structure, so this stuff isn't just extra effort for nothing.


## Nav bar

```html
<nav>
  <a href="#home">Home</a>
  <a href="#about">About</a>
  <a href="#skills">Skills</a>
  <a href="#education">Education</a>
  <a href="#contact">Contact</a>
</nav>
```

Just anchor links to the section ids for now, smooth scrolling comes later with JS.


## Skills section

```html
<section class="skills" id="skills">
  <h2>Skills</h2>
  <ul>
    <li>Python</li>
    <li>VLSI</li>
    <li>LTspice</li>
    <li>Leadership</li>
    <li>Public Speaking</li>
    <li>Project Management</li>
  </ul>
</section>
```


## Education section

```html
<section class="education" id="education">
  <h2>Education</h2>
  <p><strong>B.Tech in Electronics and Communication Engineering</strong></p>
  <p>7th Semester, currently pursuing</p>
</section>
```


## Contact form

```html
<section class="contact" id="contact">
  <h2>Contact Me</h2>
  <form>
    <label for="name">Name</label>
    <input type="text" id="name" name="name" required>

    <label for="email">Email</label>
    <input type="email" id="email" name="email" required>

    <label for="message">Message</label>
    <textarea id="message" name="message" rows="5" required></textarea>

    <button type="submit">Send</button>
  </form>
</section>
```

Form isn't connected to anything yet, just the structure for now. Might hook it up to a form service later.

# Day 4 - Styling with CSS (Typography, Colors, Spacing, Buttons, Layout)

Codomax Web Dev Internship

The page finally gets some color today. Up until now everything was plain HTML, so it's nice to actually see it start looking like a real website. Went with a simple, light/minimal theme - clean background, one accent color, nothing too loud.


## What I did today

- linked a Google Font and set up base typography
- picked a simple light color palette
- added consistent spacing/padding across sections
- styled the buttons (hero CTA, form submit)
- did a basic layout pass so sections don't feel cramped

Still all plain CSS, no Flexbox/Grid yet - that's tomorrow.


## Fonts and base styles

Used Google Fonts (Poppins) for headings and a simpler sans-serif for body text so things don't feel too heavy.

```html
<link rel="preconnect" href="https://fonts.googleapis.com">
<link href="https://fonts.googleapis.com/css2?family=Poppins:wght@400;600;700&family=Inter:wght@400;500&display=swap" rel="stylesheet">
```

```css
* {
  margin: 0;
  padding: 0;
  box-sizing: border-box;
}

body {
  font-family: 'Inter', sans-serif;
  color: #222;
  background-color: #ffffff;
  line-height: 1.6;
}

h1, h2, h3 {
  font-family: 'Poppins', sans-serif;
  font-weight: 600;
}
```


## Color palette

Kept it minimal - light background, dark text, one accent color for buttons/links so it doesn't look flat.

```css
:root {
  --bg-color: #ffffff;
  --text-color: #222222;
  --muted-text: #666666;
  --accent-color: #2f6fed;
  --section-bg: #f7f8fa;
}
```

Using CSS variables here so I don't have to hunt through the whole file if I want to change the accent color later.


## Spacing and sections

```css
section {
  padding: 60px 20px;
}

section:nth-child(even) {
  background-color: var(--section-bg);
}

h2 {
  margin-bottom: 20px;
}

p {
  margin-bottom: 12px;
  color: var(--muted-text);
}
```

Alternating section background colors slightly so each block feels separated without needing borders.


## Buttons

```css
.btn {
  display: inline-block;
  padding: 12px 28px;
  background-color: var(--accent-color);
  color: #ffffff;
  text-decoration: none;
  border-radius: 6px;
  font-weight: 500;
  transition: background-color 0.2s ease;
}

.btn:hover {
  background-color: #1f4fc4;
}

button[type="submit"] {
  padding: 10px 24px;
  background-color: var(--accent-color);
  color: #fff;
  border: none;
  border-radius: 6px;
  cursor: pointer;
}
```


## Basic layout pass

```css
.hero {
  text-align: center;
  padding: 100px 20px;
}

.about-content,
.contact form {
  max-width: 600px;
  margin: 0 auto;
}

nav {
  display: flex;
  justify-content: center;
  gap: 24px;
  padding: 16px;
}
```

Nothing responsive yet, just centering things and setting max-widths so the content doesn't stretch too wide on a big screen. Flexbox/Grid for actual responsive layout comes on Day 5.


## Things I got stuck on / figured out

- spent a while on a spacing issue where margins were behaving weird, turned out to be margin collapse between sections, fixed once I understood what was happening
- had a mismatched color at one point where the button text and background were too close in contrast, adjusted the accent color to fix it
- kept re-adjusting padding values until things felt balanced, realized consistent spacing matters more than the exact numbers

# Day 6 - Media Queries & Responsive Optimization

Codomax Web Dev Internship

Tested the site on different screen sizes today and, unsurprisingly, it looked pretty broken on mobile. Spent the day adding media queries so things adjust properly across mobile, tablet and desktop instead of just cramming the desktop layout onto a small screen.


## What I did today

- picked breakpoints: 480px (mobile), 768px (tablet), 1024px (desktop)
- collapsed the nav bar on smaller screens
- adjusted the skills grid to fewer columns on mobile/tablet
- reduced hero padding/font sizes so text doesn't overflow on small screens
- stacked the about section content vertically on mobile
- tested everything using Chrome DevTools device toolbar


## Breakpoints I'm using

```css
/* Mobile: up to 480px */
@media (max-width: 480px) { ... }

/* Tablet: up to 768px */
@media (max-width: 768px) { ... }

/* Small desktop / laptop: up to 1024px */
@media (max-width: 1024px) { ... }
```

Going with a mobile-first-ish approach but honestly wrote the desktop styles first over the past few days, so right now I'm overriding downward with max-width queries. Might switch to true mobile-first later if I have time.


## Nav bar on smaller screens

```css
@media (max-width: 768px) {
  nav {
    flex-wrap: wrap;
    gap: 12px;
    padding: 12px;
  }
}

@media (max-width: 480px) {
  nav {
    flex-direction: column;
    align-items: center;
  }
}
```

On mobile the nav just stacks vertically for now. A proper hamburger menu with JS toggle is something I might add later, but keeping it simple for now.


## Skills grid across screen sizes

```css
.skills-grid {
  display: grid;
  grid-template-columns: repeat(3, 1fr);
  gap: 16px;
}

@media (max-width: 768px) {
  .skills-grid {
    grid-template-columns: repeat(2, 1fr);
  }
}

@media (max-width: 480px) {
  .skills-grid {
    grid-template-columns: 1fr;
  }
}
```

3 columns on desktop, 2 on tablet, single column on mobile. This is the part I'm proudest of today, feels like a real responsive grid now instead of just shrinking everything.


## Hero section on mobile

```css
@media (max-width: 480px) {
  .hero h1 {
    font-size: 1.8rem;
  }

  .hero p {
    font-size: 0.95rem;
  }

  .hero {
    padding: 40px 16px;
    min-height: 50vh;
  }
}
```

Desktop font sizes looked way too big on a phone, this brings things down to a readable scale.


## About section stacking

```css
@media (max-width: 768px) {
  .about-content {
    display: flex;
    flex-direction: column;
    align-items: center;
    text-align: center;
  }

  .about-content img {
    width: 60%;
    margin-bottom: 16px;
  }
}
```


## Things I got stuck on / figured out

- had a horizontal scroll issue on mobile that I couldn't figure out for a bit, turned out one element didn't have a max-width set and was overflowing
- kept testing only by shrinking the browser window instead of using DevTools' actual device presets, switched to that and caught a few more issues
- realized some of my media query rules were conflicting because of the order I wrote them in, reordered so more specific breakpoints come after the general ones
